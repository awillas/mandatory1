## 1.2.2. The Dirichlet problem

### The dispersion coefficient $\omega$
Since
$$
\left(
c^{2}\frac{\partial^{2}}{\partial{x}^{2}}
+ c^{2}\frac{\partial^{2}}{\partial{y}^{2}}
- \frac{\partial^{2}}{\partial{t}^{2}}
\right)
\cos( k_{x} x )
\cos( k_{y} y )
\cos( \omega t )
=
-
\left(
c^{2} k_{x}^{2}
+ c^{2} k_{y}^{2}
- \omega^{2}
\right)
\cos( k_{x} x )
\cos( k_{y} y )
\cos( \omega t )
\overset{!}{=} 0,
$$
we must have
$$
c^{2} k_{x}^{2}
+ c^{2} k_{y}^{2}
- \omega^{2}
= 0.
$$
Hence,
$$
\omega = c \sqrt{ k_{x}^{2} + k_{y}^{2} }.
$$

## 1.2.3. Exact solution
A direct computation yields
$$
\left(
c^{2}\frac{\partial^{2}}{\partial{x}^{2}}
+ c^{2}\frac{\partial^{2}}{\partial{y}^{2}}
- \frac{\partial^{2}}{\partial{t}^{2}}
\right)
e^{i \left( k_{x} x + k_{y} y - \omega t \right) }
=
- (
\underbrace{
\underbrace{
c^{2} k_{x}^{2}
+ c^{2} k_{y}^{2}
}_{\omega^{2}}
- \omega^{2}
}_{0}
)
\,
e^{i \left( k_{x} x + k_{y} y - \omega t \right) }
= 0.
$$
Hence, $u(t, x, y) = e^{i \left( k_{x} x + k_{y} y - \omega t \right) }$ is an exact solution of the wave equation.

### 1.2.4 Dispersion coefficient
If $k_{x} = k_{y} = k$, the dispersion coefficient of the exact solution of the wave equation is 
$$
\omega = c\sqrt{ 2 k^{2} } = c k \sqrt{2}.
$$


Next, we compute the numerical approximation $\tilde{\omega}$ to $\omega$. Before we begin our derivation, we note the following identities:
$$
\begin{aligned}
u_{l,j}^{n} &= u_{j,l}^{n},
\\
u_{l,j}^{n+1} &= e^{-i \tilde{\omega} \Delta{t}} u_{l,j}^{n},
\\
u_{l.j+1} &= e^{i k h} u_{l,j}^{n}.
\end{aligned}
$$

For CFL number $C = 1/\sqrt{2}$,
$$
\begin{aligned}
\frac{ u_{l,j}^{n+1} - 2 u_{l,j}^{n} + u_{l,j}^{n-1} }{ \Delta{t}^{2} } 
&=
c^{2}
\left( 
\frac{ u_{l+1,j}^{n} - 2 u_{l,j}^{n} + u_{l-1,j}^{n} }{h^{2}} 
+
\frac{ u_{l,j+1}^{n} - 2 u_{l,j}^{n} + u_{l,j-1}^{n} }{h^{2}}
\right)
\\
u_{l,j}^{n+1} - 2 u_{l,j}^{n} + u_{l,j}^{n-1}
&=
\underbrace{\frac{ c^{2} \Delta{t}^{2} }{ h^{2} }}_{C^{2}}
\left( 
u_{l+1,j}^{n} - 2 u_{l,j}^{n} + u_{l-1,j}^{n}
+
u_{l,j+1}^{n} - 2 u_{l,j}^{n} + u_{l,j-1}^{n}
\right)
\\
u_{l,j}^{n+1} - 2 u_{l,j}^{n} + u_{l,j}^{n-1}
&=
\frac{ 1 }{ 2 }
\left( 
u_{l+1,j}^{n} - 2 u_{l,j}^{n} + u_{l-1,j}^{n}
+
u_{l,j+1}^{n} - 2 u_{l,j}^{n} + u_{l,j-1}^{n}
\right)
\\
u_{l,j}^{n+1} - 2 u_{l,j}^{n} + u_{l,j}^{n-1}
&=
\frac{ 1 }{ 2 }
\left( 
\underbrace{u_{l+1,j}^{n} + u_{l,j+1}^{n}}_{2 u_{l,j+1}^{n}}
- 4 u_{l,j}^{n}
+
\underbrace{u_{l-1,j}^{n} + u_{l,j-1}^{n}}_{2 u_{l,j-1}^{n}}
\right)
\\
u_{l,j}^{n+1} - 2 u_{l,j}^{n} + u_{l,j}^{n-1}
&=
u_{l,j+1}^{n} - 2 u_{l,j}^{n} + u_{l,j-1}^{n}
\\
\underbrace{ u_{l,j}^{n+1} }_{e^{-i \tilde{\omega} \Delta{t} }u_{l,j}^{n}} 
+
\underbrace{ u_{l,j}^{n-1} }_{e^{i \tilde{\omega} \Delta{t} }u_{l,j}^{n}}
&=
\underbrace{u_{l,j+1}^{n}}_{e^{i k h}u_{l,j}^{n}} 
+
\underbrace{u_{l,j-1}^{n}}_{e^{-i k h}u_{l,j}^{n}}
\\
\underbrace{\left( e^{-i \tilde{\omega} \Delta{t} } + e^{i \tilde{\omega} \Delta{t}} \right)}_{2 \cos( \tilde{\omega} \Delta{t} )}
\,
u_{l,j}^{n}
&=
\underbrace{ \left( e^{i k h} + e^{-i k h} \right)}_{2 \cos(k h) } u_{l,j}^{n}
\\
\cos( \tilde{\omega} \Delta{t} ) &= \cos(k h).
\end{aligned}
$$
Hence, there is an integer $m$, such that 
$$
\tilde{\omega} = \frac{ k h }{ \Delta{t} } + 2 \pi m
= 
k \underbrace{ \frac{ h }{ \Delta{t} } }_{c \sqrt{2}} 
+ 2 \pi m
= 
\underbrace{ c k \sqrt{2} }_{\omega}
+ 2 \pi m
= 
\omega + 2 \pi m.
$$
This means that $e^{i \tilde{\omega} t} = e^{i \omega t}$ for all real numbers $t$. Furthermore, if we are free to choose $m=0$, we obtain $\tilde{\omega} = \omega$.
